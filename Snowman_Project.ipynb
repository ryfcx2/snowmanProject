```python
import random

words = ['hello', 'cheese', 'test', 'snowman', 'person']

# shows the word and guesses left
def show_game(guess, guesses_left):
    print("word:", guess)
    print("guesses left:", guesses_left)

# gives a hint using list slicing
def give_hint(choice, guess):
    hint = choice[:2]
    print("snowman: heres a hint")
    print("snowman: first 2 letters:", hint)

    for j in range(len(hint)):
      guess[j] = hint[j]

# choose difficulty
print("snowman: choose your level")
print("1 easy")
print("2 medium")
print("3 hard")

while True:
  try:
    level = int(input("you: "))
    break
  except ValueError:
    print('snowman: please enter a valid number :(')
    continue


if level == 1:
    words = ['cat', 'dog', 'sun', 'hat', 'fish']
    guesses = 6
elif level == 2:
    words = ['hello', 'cheese', 'person', 'school', 'winter']
    guesses = 5
else:
    words = ['snowman', 'computer', 'elephant', 'mountain', 'keyboard']
    guesses = 4

# choose a random word
choice = list(random.choice(words)) #ai assisted with help using random lib

# make the hidden word
guess = ['-'] * len(choice)

points = 0

print("snowman: welcome to the game")
print("snowman: guess the word")

# give the player limited guesses
i=0
guessed_letters = []
while i<guesses:

    # show the game
    show_game(guess, guesses - i)

    # get the players guess
    n = input("you: enter a letter or type hint or enter a word ").lower()

    # check for a hint
    if n == "hint":
      give_hint(choice, guess)
      i+=1
      continue
    #check for gussed letters
    elif n in guessed_letters:
      print("snowman, you already guessed that letter")
      i+=1
      continue
    guessed_letters.append(n)


    if len(n)>1:
        if n == ''.join(choice):
            points+=30
            print("snowman: you won the game!")

            # bonus round
            bonus = input("snowman: bonus round yes or no ")

            if bonus == "yes":
                bonus_word = random.choice(words)
                bonus_choice = list(bonus_word)
                bonus_guess = ['-'] * len(bonus_choice)
                bonus_i = 0

                print("snowman: bonus round!")
                print("snowman: guess the bonus word")

                while bonus_i < guesses:
                    show_game(bonus_guess, guesses - bonus_i)

                    bonus_n = input("you: enter a letter or enter the full word (this is bonus so no hints!!!)").lower()

                    # check if they guessed the full word
                    if bonus_n == bonus_word:
                        points += 20
                        print("snowman: bonus points")
                        break

                    # check if the letter is in the word
                    if len(bonus_n) == 1:
                        if bonus_n not in bonus_choice:
                            print("snowman: wrong guess")
                            bonus_i += 1
                        else:
                            print("snowman: correct")
                            points += 5

                            for j in range(len(bonus_choice)):
                                if bonus_n == bonus_choice[j]:
                                    bonus_guess[j] = bonus_n

                            # check if they won the bonus
                            if bonus_guess == bonus_choice:
                                points += 20
                                print("snowman: bonus points")
                                break
                    else:
                        print("snowman: wrong guess, you lose sorry")
                        bonus_i += 1
                        break

                else:
                    print("snowman: no bonus points")
                    print("bonus word:", bonus_word)

                print("points:", points)

            break
        else:
            points-=100
            print("snowman: you lost the game.")
            print(f"the word was: {choice}")
            break





    # check if the letter is in the word
    if n not in choice:
        print("snowman: wrong guess")
        points -= 5
        i+=1
    else:
        print("snowman: correct")
        points += 10


    # check each letter in the word
    for j in range(len(choice)):
        if n == choice[j]:
            guess[j] = n

    # check if the player won
    if guess == choice:
        print("snowman: you won")
        print("the word was:", choice)
        print("points:", points)

        # bonus round
        bonus = input("snowman: bonus round yes or no ")

        if bonus == "yes":
            bonus_word = random.choice(words)
            bonus_choice = list(bonus_word)
            bonus_guess = ['-'] * len(bonus_choice)
            bonus_i = 0

            print("snowman: bonus round!")
            print("snowman: guess the bonus word")

            while bonus_i < guesses:
                show_game(bonus_guess, guesses - bonus_i)

                bonus_n = input("you: enter a letter or enter the full word (this is bonus round, no hints!!!) ").lower()

                # check if they guessed the full word
                if bonus_n == bonus_word:
                    points += 20
                    print("snowman: bonus points")
                    break

                # check if the letter is in the word
                if len(bonus_n) == 1:
                    if bonus_n not in bonus_choice:
                        print("snowman: wrong guess")
                        bonus_i += 1
                    else:
                        print("snowman: correct")
                        points += 5

                        for j in range(len(bonus_choice)):
                            if bonus_n == bonus_choice[j]:
                                bonus_guess[j] = bonus_n

                        # check if they won the bonus
                        if bonus_guess == bonus_choice:
                            points += 20
                            print("snowman: bonus points")
                            break
                else:
                    print("snowman: wrong guess")
                    bonus_i += 1

            else:
                print("snowman: no bonus points")
                print("bonus word:", bonus_word)

            print("points:", points)

        break

# if all guesses are used
else:
    print("snowman: you lost")
    print("the word was:", choice)
    print("points:", points)
```
